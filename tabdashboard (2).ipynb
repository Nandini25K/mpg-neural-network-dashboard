```python
import pandas as pd
import numpy as np

df = pd.read_csv("auto_mpg.csv")


print(df.isnull().sum())

df['horsepower'].fillna(df['horsepower'].mean(), inplace=True)
```

```python

df['power_to_weight'] = df['horsepower'] / df['weight']
df['efficiency_factor'] = df['horsepower'] / df['displacement']

df = pd.get_dummies(df, columns=['origin'], prefix='origin')
```

```python
from sklearn.preprocessing import StandardScaler


features = ['cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'power_to_weight', 'efficiency_factor', 'origin_1', 'origin_2', 'origin_3']
target = 'mpg'


from sklearn.model_selection import train_test_split

X = df[features]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense


model = Sequential([
    Dense(64, activation='relu', input_shape=(len(features),)),
    Dense(64, activation='relu'),
    Dense(1, activation='linear')  # Output layer for regression
])


model.compile(optimizer='adam', loss='mse', metrics=['mae'])


model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=16,
    validation_data=(X_test, y_test)
)
```

```python

df['predicted_mpg'] = model.predict(scaler.transform(df[features])).flatten()


df[['car_name', 'model_year', 'mpg', 'predicted_mpg']].to_csv("mpg_predictions.csv", index=False)
```
